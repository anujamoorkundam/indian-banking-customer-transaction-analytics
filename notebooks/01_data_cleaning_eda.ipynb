```python
bins = [0, 18, 25, 35, 45, 55, 65, float('inf')]
```

```python
labels = [
    'Under 18',
    '18-24',
    '25-34',
    '35-44',
    '45-54',
    '55-64',
    '65+'
]
```

```python
df['AgeGroup'] = pd.cut(
    df['CustomerAge'],
    bins=bins,
    labels=labels,
    right=False
)
```

```python
def get_time_of_day(hour):
    if 6 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'
```

```python
df['TimeOfDay'] = df['TransactionHour'].apply(get_time_of_day)
```

```python
df[['TransactionTime', 'TransactionHour', 'TimeOfDay']].head(20)
```

```python
amount_bins = [-0.01, 0.01, 500, 1000, 5000, 10000, 50000, float('inf')]

amount_labels = [
    'Zero Amount',
    'Below 500',
    '500-999',
    '1K-4.9K',
    '5K-9.9K',
    '10K-49.9K',
    '50K+'
]
```
